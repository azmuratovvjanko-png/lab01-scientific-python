```python
import sympy as sp

x = sp.symbols('x')

g = x**4 - 5*x**2 + 4

print("Факторизация:", sp.factor(g))
print("Корни:", sp.solve(sp.Eq(g, 0), x))
print("Производная:", sp.diff(g, x))
print("Интеграл:", sp.integrate(g, x))

exact_value = sp.simplify(g.subs(x, sp.sqrt(2)))
approx_value = sp.N(exact_value, 12)
print("Точное значение:", exact_value)
print("Приближённое значение:", approx_value)
```

### Түсіндірме
float сандары екілік жүйеде сақталады, сондықтан 0.1 мен 0.2 дәл көрсетілмейді. Нәтижесінде 0.1 + 0.2 саны 0.3-тен өте аз шамаға ерекшеленеді, және == салыстыруы False береді. math.isclose рұқсат етілген қателікті ескереді, сондықтан True береді.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

x = sp.symbols('x')
g = x**4 - 5*x**2 + 4

g_num = sp.lambdify(x, g, 'numpy')
x_values = np.linspace(-3, 3, 401)
y_values = g_num(x_values)
```

```python
x_coarse = np.linspace(-3, 3, 21)
y_coarse = g_num(x_coarse)

plt.figure(figsize=(8, 4.5))
plt.plot(x_values, y_values, label='401 нүкте')
plt.plot(x_coarse, y_coarse, 'o--', label='21 нүкте')
plt.axhline(0, color='black', linewidth=0.8)
plt.grid(alpha=0.3)
plt.legend()
plt.title('Тор тығыздығын салыстыру')
plt.show()
```

### Тор тығыздығының әсері
401 нүктелі график тегіс қисық береді. 21 нүктелі графикте нүктелер аз болғандықтан қисық сынық сызыққа ұқсайды, ал тамырлар мен минимум нүктелерінің орны дәл көрінбейді. Демек, нүкте неғұрлым көп болса, график функцияның нақты пішініне соғұрлым жақын болады.

```python
f2 = x**4 - 10*x**2 + 9

print('Факторизация:', sp.factor(f2))
print('Тамырлар:', sp.solve(sp.Eq(f2, 0), x))
print('Туынды:', sp.diff(f2, x))
print('Интеграл:', sp.integrate(f2, x))

f2_num = sp.lambdify(x, f2, 'numpy')
xs = np.linspace(-4, 4, 401)   # диапазонды кестеден алыңыз
ys = f2_num(xs)

roots2 = [float(r) for r in sp.solve(sp.Eq(f2, 0), x)]

plt.figure(figsize=(8, 4.5))
plt.plot(xs, ys, label='f(x)')
plt.plot(roots2, [0]*len(roots2), 'ro', label='Тамырлар')
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(alpha=0.3)
plt.legend()
plt.title('Жеке тапсырма: график f(x)')
plt.show()
```
